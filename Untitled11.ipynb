```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from scipy.spatial.distance import pdist

df = pd.read_csv("temperatures.csv",sep=";")
df.head()
```

```python
months = [
"Janvier", "Fevrier", "Mars", "Avril", "Mai", "Juin",
"Juillet", "Aout", "Septembre", "Octobre", "Novembre", "Decembre"
]
X = df[months]
print(X.shape)
scaler =StandardScaler()
X_scaled = scaler.fit_transform(X)
# Vérification : moyenne ≈ 0 et écart-type ≈ 1
print("mean",X_scaled.mean(axis=0))

print("std",X_scaled.std(axis=0))
```

```python
inertias = []
# Valeurs de k testées
K_values = range(1, 8)
for k in K_values:
 # Création du modèle k-means
 kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
 # Apprentissage du modèle
 kmeans.fit(X_scaled)
 # Stockage de l'inertie intra-classe
 inertias.append(kmeans.inertia_)
```

```python
plt.figure()
plt.plot(K_values, inertias, marker='o')
plt.xlabel("Nombre de clusters k")
plt.ylabel("Inertie intra-classe")
plt.title("Méthode du coude – K-means")
plt.show()
```

```python
k_opt = 3
# Application finale de k-means
kmeans = KMeans(n_clusters=k_opt, random_state=42, n_init=10)
clusters_kmeans = kmeans.fit_predict(X_scaled)
df["Cluster_KMeans"] = clusters_kmeans
plt.figure()
plt.scatter(
df["Longitude"],
df["Latitude"],
c=df["Cluster_KMeans"],
cmap="tab10",
s=80
)
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Clustering K-means – Représentation géographique")
plt.colorbar(label="Cluster")
plt.show()
```

```python
# Q10. Calcul de la matrice de distances
# Calcul de la matrice des distances euclidiennes
distance_matrix = pdist(X_scaled, metric="euclidean")
# Q11. Application de la CAH (méthode de Ward)
# Construction de la hiérarchie avec la méthode de Ward
Z = linkage(distance_matrix, method="ward")
# Q12. Tracé du dendrogramme
plt.figure(figsize=(12, 5))
dendrogram(Z, labels=df["City"].values, leaf_rotation=90)
plt.xlabel("Villes")
plt.ylabel("Distance")
plt.title("Dendrogramme – CAH (méthode de Ward)")
plt.show()
```

```python
# Q13. Découpage de l’arbre et affectation aux clusters
# Nombre de clusters choisi
k_cah = 3
# Découpage du dendrogramme
clusters_cah = fcluster(Z, k_cah, criterion="maxclust")
# Ajout au DataFrame
df["Cluster_CAH"] = clusters_cah
# threshold = 7
# clusters_cah = fcluster(Z, threshold, criterion="distance")
# Représentation géographique des clusters (CAH)
plt.figure()
plt.scatter(
df["Longitude"],
df["Latitude"],
c=df["Cluster_CAH"],
cmap="tab10",
s=80
)
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("CAH – Représentation géographique")
plt.colorbar(label="Cluster")
plt.show()
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from matplotlib.image import imread

# ---------- Q17 ----------
# 17)a)on transform limage de pixel en un tableau pour kmeans peut lire  l'interpretez  facilement
# b)Une ligne de ce tableau représente un pixel unique de l'image rouge bleu green
# c) car chque pixel est presente par 3 couleur statistique
FICHIER = "visage.bmp"   # mets ton image dans le même dossier que ce script

if not os.path.exists(FICHIER):
    print("Image introuvable !")
    exit()

image = imread(FICHIER)

# on garde seulement R, V, B (supprime le canal alpha s'il existe)
image = image[:, :, :3]
h, w, c = image.shape
print("Dimensions de l'image originale :", image.shape)

# a) aplatir : (h, w, 3) -> (h*w, 3)
image_array = image.reshape(h * w, c)
print("Dimensions du tableau de pixels  :", image_array.shape)

# b) c) lignes = pixels, colonnes = R, V, B
print("10 premiers pixels (Rouge, Vert, Bleu) :")
print(image_array[:10])


# ---------- Q18 ----------
# q18 a)cest definir le nombre total de couleur a letat final
# b) une class  represente  les pixels qui sont les memes et pour le centre de k cest lui qui somme tout les clour et retourn la moyenne de la couleur
# cest la moyenne de group de pixel  et il est meme (r,g,b) et il cree des classes pour reduire les couleurs
k = 16
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
kmeans.fit(image_array)

palette_couleurs = kmeans.cluster_centers_
print("\nPalette des", k, "couleurs (centres) :")
print(palette_couleurs.round().astype(int))


# ---------- Q19 ----------
#q19 a) il va changer sa couleur et sadapter a la couleur qui appartient au plus
# b)quand replace chaque pixel par le k ou bien la classe qui appertient a .on va reduire beaucoup le stockage  dans la memory a cause de ce principe
# c)Pour chaque pixel on lit son numero de classe on recupere la couleur du centre  puis on remet le tableau (h×w, 3) sous la forme (h, w, 3) pour afficher l’image
compressed_pixels = palette_couleurs[kmeans.labels_]
compressed_image = compressed_pixels.reshape(h, w, c)
compressed_image = compressed_image.astype(np.uint8)

print("\nCouleurs distinctes avant :", len(np.unique(image_array, axis=0)))
print("Couleurs distinctes après :", len(np.unique(compressed_pixels, axis=0)))
print("Bits par pixel avant : 24")
print("Bits par pixel après :", int(np.ceil(np.log2(k))))

# affichage original vs compressée
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Image originale")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(compressed_image)
plt.title("Image compressée avec K = " + str(k))
plt.axis("off")

plt.show()


# ---------- Q20 ----------
# Q20 a)la qualite devient faible et ils sont fusionnes
# b)La qualité est tres bonne
# c)car si on veut un meilleure qualite donc on a besoin de plus de couleur donc il vas consommer plus et vis versa
ks = [2, 4, 16, 64]

plt.figure(figsize=(15, 4))

plt.subplot(1, 5, 1)
plt.imshow(image)
plt.title("Original")
plt.axis("off")

print("\nK  -> bits par pixel")
for i in range(4):
    kk = ks[i]
    km = KMeans(n_clusters=kk, random_state=42, n_init=1)
    km.fit(image_array)
    img_k = km.cluster_centers_[km.labels_].reshape(h, w, c).astype(np.uint8)

    print(kk, "->", int(np.ceil(np.log2(kk))), "bits")

    plt.subplot(1, 5, i + 2)
    plt.imshow(img_k)
    plt.title("K = " + str(kk))
    plt.axis("off")

plt.show()
inertias = []
K_values = range(1, 21)

for k_val in K_values:
    kmeans_test = KMeans(n_clusters=k_val, random_state=42, n_init=10)
    kmeans_test.fit(image_array)
    inertias.append(kmeans_test.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K_values, inertias, marker='o', color='b')
plt.xlabel("Nombre de couleurs (K)")
plt.ylabel("Inertie intra-classe")
plt.title("Méthode du coude - Choix optimal de K pour l'image")
plt.show()
```
